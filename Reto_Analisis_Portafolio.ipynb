```python
# ============================================
# RETO INDIVIDUAL
# Reemplaza estos datos por los asignados

# Tus datos del reto
# Activo Cantidad Precio de compra Precio actual
# NVDA 18 $111 $139
# AAPL 10 $239 $302
# GOOGL 14 $290 $344
# AMZN 18 $303 $313

# ============================================

# Paso 1: Definir los datos de tu portafolio y precios actuales.
# Se utiliza un diccionario para organizar la información de cada activo:
# cantidad comprada, precio de compra y precio actual de mercado.
portafolio = {
    "NVDA": {"cantidad": 18, "precio_compra": 111, "precio_actual": 139},
    "AAPL": {"cantidad": 10, "precio_compra": 239, "precio_actual": 302},
    "GOOGL": {"cantidad": 14, "precio_compra": 290, "precio_actual": 344},
    "AMZN": {"cantidad": 18, "precio_compra": 303, "precio_actual": 313}
}

# Paso 2: Crear una lista con los nombres de los activos para iterar sobre ellos.
# Esto facilita el procesamiento de cada activo en un ciclo.
activos = list(portafolio.keys())

# Paso 3: Definir la función para analizar un activo individual.
# Esta función encapsula la lógica de cálculo para valor invertido, valor actual,
# ganancia y rendimiento, promoviendo la reutilización de código.
def analizar_activo(cantidad, precio_compra, precio_actual):
    valor_invertido = cantidad * precio_compra
    valor_actual = cantidad * precio_actual
    ganancia = valor_actual - valor_invertido
    rendimiento = (ganancia / valor_invertido) * 100
    return valor_invertido, valor_actual, ganancia, rendimiento

# Paso 4: Inicializar una lista para almacenar los resultados de cada activo.
# Esta lista de diccionarios será la base para el informe final.
resultados = []

# Paso 5: Iterar sobre cada activo del portafolio para realizar el análisis completo.
# Dentro de este ciclo, se obtienen los datos, se llama a la función `analizar_activo`,
# se determina la recomendación y se guarda todo en la lista `resultados`.
for activo in activos:
    cantidad = portafolio[activo]["cantidad"]
    precio_compra = portafolio[activo]["precio_compra"]
    precio_actual = portafolio[activo]["precio_actual"]

    # Usar la función para obtener los cálculos financieros.
    valor_invertido, valor_actual, ganancia, rendimiento = analizar_activo(
        cantidad, precio_compra, precio_actual
    )

    # Determinar la recomendación de inversión basada en el rendimiento.
    if rendimiento >= 10:
        recomendacion = "MANTENER"
    elif rendimiento >= 0:
        recomendacion = "REVISAR"
    else:
        recomendacion = "ANALIZAR VENTA"

    # Agregar todos los datos calculados para el activo a la lista de resultados.
    resultados.append({
        "activo": activo,
        "valor_invertido": valor_invertido,
        "valor_actual": valor_actual,
        "ganancia": ganancia,
        "rendimiento": rendimiento,
        "recomendacion": recomendacion
    })

# Paso 6: Generar el informe resumido del portafolio.
# Se calculan los totales y se imprime un formato legible para el analista.
total_invertido = 0
total_actual = 0

print("=" * 65)
print("ANÁLISIS DEL PORTAFOLIO")
print("=" * 65)

# Imprimir los detalles de cada activo y sumar los totales.
for resultado in resultados:
    total_invertido += resultado["valor_invertido"]
    total_actual += resultado["valor_actual"]

    print(
        f"{resultado['activo']:6} | "
        f"Actual: ${resultado['valor_actual']:,.0f} | "
        f"Rendimiento: {resultado['rendimiento']:.2f}% | "
        f"{resultado['recomendacion']}"
    )

# Calcular la ganancia y el rendimiento total del portafolio.
ganancia_total = total_actual - total_invertido
rendimiento_total = (ganancia_total / total_invertido) * 100

# Imprimir el resumen total del portafolio.
print("=" * 65)
print(f"Total invertido: ${total_invertido:,.0f}")
print(f"Valor actual:    ${total_actual:,.0f}")
print(f"Ganancia/pérdida: ${ganancia_total:,.0f}")
print(f"Rendimiento:     {rendimiento_total:.2f}%")

# Paso 7: Identificar y mostrar el activo con el mejor rendimiento.
# Se utiliza la función `max` con una función `lambda` para encontrar
# el diccionario en `resultados` con el valor de 'rendimiento' más alto.
mejor_activo = max(resultados, key=lambda x: x["rendimiento"])

print("=" * 65)
print("Activo con mejor rendimiento:")
print(f"  Activo: {mejor_activo['activo']}")
print(f"  Rendimiento: {mejor_activo['rendimiento']:.2f}%")
print(f"  Recomendación: {mejor_activo['recomendacion']}")
print("=" * 65)
```
